```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd

df = pd.read_csv("cleaned_data/information/panel_macro_cleaned.csv")
# 1. Lọc dữ liệu cho các quốc gia trọng tâm
focus_countries = ['SGP', 'USA', 'IND', 'DEU', 'CHN', 'JPN', 'VNM', 'THA']
df_plot = df[df['Country'].isin(focus_countries)].copy()

# 2. Vẽ biểu đồ đường cơ bản
fig_inflation = px.line(
    df_plot,
    x='Year',
    y='Inflation_CPI_Pct',
    color='Country',
    title="Biến động Lạm phát (CPI) và Các cú sốc kinh tế toàn cầu (1990-2024)",
    labels={
        'Year': 'Năm', 
        'Inflation_CPI_Pct': 'Tỷ lệ Lạm phát (%)', 
        'Country': 'Quốc gia'
    }
)

# 3. Định nghĩa các mốc khủng hoảng
crises = {
    2008: "Khủng hoảng Tài chính",
    2018: "Thương chiến Mỹ-Trung",
    2020: "Đại dịch COVID-19",
    2022: "Khủng hoảng Năng lượng"
}

# 4. Thêm các đường đánh dấu sự kiện (Vertical Lines)
for year, title in crises.items():
    fig_inflation.add_vline(
        x=year, 
        line_width=2, 
        line_dash="dash", 
        line_color="red", 
        opacity=0.7,
        annotation_text=f"<b>{title}</b>", 
        annotation_position="top left",
        annotation_font=dict(color="red", size=11)
    )

# 5. Tùy chỉnh giao diện
fig_inflation.update_traces(line=dict(width=2.5))
fig_inflation.update_layout(
    template="plotly_white",
    height=700,
    margin=dict(l=60, r=40, t=80, b=60),
    hovermode="x unified", # Giúp hiển thị lạm phát của tất cả các nước tại cùng 1 năm khi di chuột
    xaxis=dict(showgrid=True, gridcolor='lightgray'),
    yaxis=dict(showgrid=True, gridcolor='lightgray', zeroline=True, zerolinecolor='black')
)

# Đưa đường của Việt Nam lên trên cùng và đổi màu nổi bật (Tùy chọn)
for i, trace in enumerate(fig_inflation.data):
    if trace.name == 'VNM':
        fig_inflation.data[i].line.color = '#E60000' # Đỏ đậm
        fig_inflation.data[i].line.width = 4
    elif trace.name == 'USA':
        fig_inflation.data[i].line.color = '#000080' # Xanh than

fig_inflation.show(config={'responsive': True})
fig_inflation.write_html("line_chart_inflation_crises.html")
```

```python
import plotly.express as px
import plotly.graph_objects as go

# 1. Lọc dữ liệu cho các quốc gia trọng tâm (Sử dụng chung df_plot từ trước)
focus_countries = ['SGP', 'USA', 'IND', 'DEU', 'CHN', 'JPN', 'VNM', 'THA']
df_plot = df[df['Country'].isin(focus_countries)].copy()

# 2. Vẽ biểu đồ đường cho Tăng trưởng GDP
fig_gdp = px.line(
    df_plot,
    x='Year',
    y='GDP_Growth_Pct',
    color='Country',
    title="Tốc độ Tăng trưởng GDP và Sức chống chịu trước Khủng hoảng (1990-2024)",
    labels={
        'Year': 'Năm', 
        'GDP_Growth_Pct': 'Tăng trưởng GDP (%)', 
        'Country': 'Quốc gia'
    }
)

# 3. Định nghĩa các mốc khủng hoảng
crises = {
    2008: "Khủng hoảng Tài chính",
    2018: "Thương chiến Mỹ-Trung",
    2020: "Đại dịch COVID-19",
    2022: "Khủng hoảng Năng lượng"
}

# 4. Thêm các đường đánh dấu sự kiện
for year, title in crises.items():
    fig_gdp.add_vline(
        x=year, 
        line_width=2, 
        line_dash="dash", 
        line_color="red", 
        opacity=0.7,
        annotation_text=f"<b>{title}</b>", 
        annotation_position="bottom left", # Đưa chữ xuống dưới để không đè lên phần lạm phát nếu ghép chart
        annotation_font=dict(color="red", size=11)
    )

# Thêm đường cơ sở 0% (Ranh giới Suy thoái kinh tế)
fig_gdp.add_hline(y=0, line_width=2, line_dash="solid", line_color="black", opacity=0.8)

# 5. Tùy chỉnh giao diện và làm nổi bật Việt Nam
fig_gdp.update_traces(line=dict(width=2.5))
fig_gdp.update_layout(
    template="plotly_white",
    height=700,
    margin=dict(l=60, r=40, t=80, b=60),
    hovermode="x unified",
    xaxis=dict(showgrid=True, gridcolor='lightgray'),
    yaxis=dict(showgrid=True, gridcolor='lightgray', title="Tăng trưởng GDP (%)")
)

for i, trace in enumerate(fig_gdp.data):
    if trace.name == 'VNM':
        fig_gdp.data[i].line.color = '#E60000' # Đỏ đậm
        fig_gdp.data[i].line.width = 4
    elif trace.name == 'USA':
        fig_gdp.data[i].line.color = '#000080' # Xanh than
    elif trace.name == 'DEU':
        fig_gdp.data[i].line.color = '#808080' # Xám cho Đức (đại diện xuất khẩu bị tổn thương nặng 2020)

fig_gdp.show(config={'responsive': True})
fig_gdp.write_html("line_chart_gdp_growth_crises.html")
```
